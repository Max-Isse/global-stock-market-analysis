```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("../data/stock_data.csv", index_col=0)

returns = data.pct_change()

print("Average Returns")
print(returns.mean())

# Plot price trends
data.plot(figsize=(12,6))
plt.title("Stock Price Trends")
plt.xlabel("Date")
plt.ylabel("Price")
plt.savefig("../visuals/price_trends.png")
plt.show()
```

## Key Insights
* Technology stocks showed strong long-term growth.

* Tesla exhibited the highest volatility.

* Apple and Microsoft demonstrated consistent price appreciation.
